```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("dataset/airbnb_listings.csv")
df.head()
```

```python
print(df.columns.tolist())
print(df[['price', 'name']].isnull().sum() if 'price' in df.columns else df.isnull().sum())
```

```python
df['High Price Indicator'] = df['price'] > 100
df[['price', 'High Price Indicator']].head()
```

```python
def categorize(p):
    if pd.isna(p):
        return 'Unknown'
    if p < 100:
        return 'Budget'
    elif p < 200:
        return 'Mid-Range'
    elif p < 400:
        return 'Premium'
    return 'Luxury'

df['Property Category'] = df['price'].apply(categorize)
df['Property Category'].value_counts()
```

```python
reviews = df['number_of_reviews'].replace(0, np.nan)
df['Price Per Review'] = df['price'] / reviews
df[['price', 'number_of_reviews', 'Price Per Review']].head()
```

```python
avail = df['availability_365'].fillna(0)
scores = np.ceil((avail / 365.0) * 5)
df['Availability Score'] = np.clip(scores, 1, 5).astype(int)
df['Availability Score'].value_counts().sort_index()
```

```python
df.to_csv("airbnb_listings_enriched.csv", index=False)
```
